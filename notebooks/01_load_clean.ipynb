```python
# ==================================================
# Configuració del projecte
# ==================================================

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
# ==================================================
# Imports
# ==================================================

import pandas as pd

from src.config import (
    DATA_RAW,
    DATA_INTERIM,
)

from src.utils_io import (
    load_csv,
    save_csv,
)

from src.cleaning import (
    clean_numeric_columns,
    replace_missing,
    standardize_column_names,
)

from src.cleaning import fix_tipus_territori
```

```python
# ==================================================
# Warnings
# ==================================================

import warnings

warnings.filterwarnings("ignore")
```

```python
# ==================================================
# Funcions auxiliars
# ==================================================

def basic_cleaning(df):
    """
    Neteja estàndard aplicada a tots els datasets.
    """

    df = standardize_column_names(df)

    df = fix_tipus_territori(df)

    df = replace_missing(df)

    return df
```

```python
# ==================================================
# Carregar datasets RAW
# ==================================================

print("Carregant datasets RAW...")

df_rent = load_csv(
    DATA_RAW / "Preu lloguer.csv"
)

df_income = load_csv(
    DATA_RAW / "Renda_Barri.csv"
)

df_nat = load_csv(
    DATA_RAW / "Poblacio nacionalitat.csv"
)

df_pop = load_csv(
    DATA_RAW / "Poblacio.csv"
)

df_atur = load_csv(
    DATA_RAW / "atur.csv"
)

df_est = load_csv(
    DATA_RAW / "nivell escolar.csv"
)

df_cont = load_csv(
    DATA_RAW / "num_contractes lloguer.csv"
)

df_sup = load_csv(
    DATA_RAW / "superficie mitjana.csv"
)

df_hut = load_csv(
    DATA_RAW / "us turistic.csv"
)
```

```python
# ==================================================
# Dataset renda
# ==================================================

df_income = df_income[
    df_income["areaType"] == "neighborhood"
]

df_income["areaType"] = (
    df_income["areaType"]
    .replace(
        "neighborhood",
        "Barri"
    )
)

df_income = df_income.rename(
    columns={
        "areaType": "tipus_de_territori",
        "areaLabel": "territori",
    }
)

df_income = basic_cleaning(df_income)
```

```python
# ==================================================
# Dataset lloguer
# ==================================================

df_rent = clean_numeric_columns(df_rent)

df_rent = basic_cleaning(df_rent)

df_rent = df_rent[
    df_rent["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Dataset població
# ==================================================

df_pop = basic_cleaning(df_pop)

df_pop = df_pop[
    df_pop["tipus_de_territori"] == "Barri"
]

population_cols = [
    c for c in df_pop.columns
    if c.startswith("01_gen._")
]

df_pop = df_pop[
    population_cols
    + ["territori", "tipus_de_territori"]
]
```

```python
# ==================================================
# Dataset atur
# ==================================================

df_atur = basic_cleaning(df_atur)

df_atur = df_atur[
    df_atur["tipus_de_territori"] == "Barri"
]

atur_cols = [
    c for c in df_atur.columns
    if c.startswith("31_gen._")
]

df_atur = df_atur[
    atur_cols
    + ["territori", "tipus_de_territori"]
]
```

```python
# ==================================================
# Dataset nacionalitat
# ==================================================

df_nat = basic_cleaning(df_nat)

df_nat = df_nat[
    df_nat["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Dataset nivell estudi
# ==================================================

df_est = basic_cleaning(df_est)

df_est = df_est[
    df_est["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Dataset numero de contractes
# ==================================================

df_cont = basic_cleaning(df_cont)

df_cont = df_cont[
    df_cont["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Dataset superficie
# ==================================================

df_sup = basic_cleaning(df_sup)

df_sup = df_sup[
    df_sup["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Dataset habitatge d'us turistic
# ==================================================

df_hut = basic_cleaning(df_hut)

df_hut = df_hut[
    df_hut["tipus_de_territori"] == "Barri"
]
```

```python
# ==================================================
# Exportació datasets clean
# ==================================================

print("Guardant datasets INTERIM...")

datasets = {
    "rent_clean.csv": df_rent,
    "income_clean.csv": df_income,
    "nationality_clean.csv": df_nat,
    "population_clean.csv": df_pop,
    "atur_clean.csv": df_atur,
    "estudis_clean.csv": df_est,
    "contractes_clean.csv": df_cont,
    "superficie_clean.csv": df_sup,
    "turisme_clean.csv": df_hut,
}

for filename, dataset in datasets.items():

    save_csv(
        dataset,
        DATA_INTERIM / filename,
    )
```
